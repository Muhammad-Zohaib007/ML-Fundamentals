```python
import numpy as np 
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
# Data is Loaded

df = pd.read_csv(r'Titanic-Dataset.csv')
df.head()
```

```python
df['Age']                # one column -> Series
df[['Age', 'Sex']]       # multiple columns -> DataFrame
```

```python
df.groupby(['Pclass', 'Sex'])['Survived'].mean()
```

```python
df.isnull().sum()
df.isnull().sum() / len(df) * 100 # percentage missing per column
```

```python
sns.heatmap(df.isnull(), cbar=False)
```

```python
df.dropna(subset=['Embarked'])
df.drop('Cabin', axis = 1)
```

```python
df['Age'].fillna(df['Age'].median(), inplace=True)
df['Embarked'].fillna(df['Embarked'].mode()[0], inplace=True)
```

```python
plt.hist(df['Age'], bins=20, color='skyblue', edgecolor='black')
plt.xlabel('Age')
plt.ylabel('Frequency', size=20)
plt.title('Age Distribution', size = 20)
```

```python
df['Pclass'].value_counts().plot(kind='bar', color='grey')
plt.xlabel('Passenger Class')
plt.ylabel('Count')
plt.title('Passengers per Class')
plt.show()
```

```python
plt.scatter(df['Age'], df['Fare'], alpha=0.5)
plt.xlabel('Age', size=20)
plt.ylabel('Fare', size=20)
plt.title('Age vs Fare', size=20)
plt.show()
```

```python
sns.boxplot(x='Pclass', y='Fare', data=df)
plt.title('Fare Distribution by Class')
plt.show()
```

```python
numeric_df = df.select_dtypes(include=['int64', 'float64'])
sns.heatmap(numeric_df.corr(), annot=True, cmap='coolwarm', fmt='.2f')
plt.title('Correlation Heatmap')
plt.show()
```
